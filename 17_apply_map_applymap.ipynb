### **Apply Functions in Pandas**

*map()*, *apply()*, and *applymap()* are all used to transform data in pandas, but they work on different objects.

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Method</th>
<th>Works on</th>
<th>Purpose</th>
</tr>
<tr>
<td><i>map()</i></td>
<td>Single Series (one column)</td>
<td>Replace or transform values one by one</td>
</tr>
<tr>
<td><i>apply()</i></td>
<td>Series or DataFrame</td>
<td>Apply a function to rows, columns, or a Series</td>
</tr>
<tr>
<td><i>applymap()</i></td>
<td>Entire DataFrame</td>
<td>Apply a function to every individual cell (deprecated in recent pandas versions; use DataFrame.map() instead if available)</td>
</tr>
</table>

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Students.csv')
df.head()
````

#### 1. ***map()***

Works only on a Series (single column).

**Syntax**
```python
Series.map(arg)
```

**arg can be:**

- Dictionary
- Function
- Series

**Example 1: Dictionary Mapping**(Add State from city)


````python
state_code = {
    'Vijayawada': 'AP',
    'Visakhapatnam': 'AP',
    'Pune': 'MH',
    'Delhi':'DH',
    'Chennai':'TN',
    'Mumbai':'MH',
    'Bengaluru':'KA',
    'Hyderabad': 'TS'
}
df['State_code'] = df['City'].map(state_code)
df.head()
````

**Example 2: Convert Gender to short form**

````python
df['Gender'] = df['Gender'].map({
    'M': 'Male',
    'F':'Female'
})
df
````

**Example 3: String Length**

````python
df['Name_Len'] = df['Name'].map(lambda x : len(x))
df
````

**When to use map()**

- Replace values
- Create labels
- Transform one column

#### 2. ***apply()***

apply() is much more flexible.

**It Works on:**

- Series
- DataFrame

**Syntax:**
```python
Series.apply(function)
#or
DataFrame.apply(function, axis=0)

```
- axis=1 means work row by row.

- axis=0 means work column by column.

**Example 1: Calculate Total Marks**

````python
df['Total'] = df[['Math', 'Science', 'English']].apply(sum, axis=1)
df[['Name', 'Math', 'Science', 'English', 'Total']].head()
````

**Example 2: Grades based on total scores**

````python
df['Grade'] = df['Total'].apply(
    lambda x: 'A' if x >= 270
    else 'B' if x >= 240
    else 'C'
)
df[['Name', 'Math', 'Science', 'English', 'Total', 'Grade']].head()
````

**Example 3: Highest mark in each student**

````python
df['Highest'] = df[['Math', 'Science', 'English']].apply(max, axis=1)
df[['Name', 'Math','Science', 'English', 'Highest']].head()
````

**Example 4: Calculate Average**

````python
df['Average'] = df[['Math', 'Science', 'English']].apply(lambda x: x.mean(), axis=1)
df[['Name', 'Math','Science', 'English', 'Average']].head()
````

#### 3. ***applymap()* (Element-wise)**

*applymap()* applies a function to every individual cell in a DataFrame.

> Note: In recent versions of pandas, DataFrame.applymap() is deprecated. Use DataFrame.map() for element-wise operations instead.

**Example 1: Add 10 Marks to Every Subject**

````python
print(df[['Math', 'Science', 'English']].head())
df[['Math', 'Science', 'English']].map(lambda x : x + 10).head()
````

**Example 2: Multiply Every Mark by 2**

````python
print(df[['Math', 'Science', 'English']].head())
df[['Math', 'Science', 'English']].map(lambda x : x * 2).head()
````

**Example 3: Convert Numbers to Strings**

````python
print(df[['Math', 'Science', 'English']].head())
df[['Math', 'Science', 'English']].map(lambda x : f'{x} Marks').head()
````

Easy way to remember
- map() → One column (Series).
- apply() → Rows, columns, or a Series.
- df.map() → Every cell in a DataFrame.
