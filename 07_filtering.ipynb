### **Filtering in Pandas**

Filtering is the process of selecting only the rows (or sometimes columns) from a DataFrame that satisfy one or more conditions.

```python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
```

```python
df.columns
```

#### 1. **Filter Rows by a Single Condition**

```python
df[df['Department'] == 'IT']
```

#### 2. **Filter with Multiple Conditions**

Important Notes
- Use **&** instead of **and**.
- Use **|** instead of **or**.

##### **AND (&)**

Employees older than 28 and salary above 55000:

```python
df[(df['Age'] > 28) & (df['Salary'] > 55000) ]
```

##### **OR (|)**

Employees younger than 26 or salary above 65000:

```python
df[(df['Age'] > 26) | (df['Salary'] > 65000)]
```

#### 3. **Filter Using *isin()***

```python
dep = ['IT', 'Operations']
df[df['Department'].isin(dep)]
```

#### 4. **Filter Missing Values**

```python
print(df[df['Experience'].isna()]) # Missing Values
print()
df[df['Experience'].notna()]
```

#### 5. **Filter with String Conditions**

Names starting with "A":

```python
df[df['Name'].str.startswith('A')]
```

Names containing "a":

```python
df[df['Name'].str.contains('t', case='False')]
```

#### 6. **Using query()**

This is often cleaner for complex conditions.

```python
df.query('Age > 30 and Department == "IT"')
```

#### 7. **Filter Columns**

Select specific columns:

```python
df[['Name', 'Salary']]
```

#### 8. **Using loc[]**

loc[] filters rows and can also select columns at the same time.

```python
df.loc[df['Age'] > 50, ['Name', 'Age', 'Salary']]
```

#### 9. **Between()**

```python
df[df['Age'].between(25,30)]
```

#### 10. **where()**

Keeps values where condition is True.

False values become NaN.

```python
df.where(df['Department'] == 'HR')
```

**Replace False values**

```python
df.where(df['Department'] == 'HR', 'N/A')
```

#### 11. **mask()**

Opposite of where().

Condition True → replaced.

Condition False → kept.

```python
df.mask(df['Department'] == 'HR')
```

**Replace**

```python
df.mask(df['Department'] == 'HR', 'N/A')
```
