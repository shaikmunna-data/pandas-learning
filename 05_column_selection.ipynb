### **Column Selection in Pandas**

Column selection means choosing one or more columns from a DataFrame.

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
df
````

#### 1. **Select a Single Column**

**Method 1:** Bracket Notation (Recommended)

````python
df['Name']
````

**Method 2:** Dot Notation

````python
df.Name
````

⚠️ Use only when:

- Column name has no spaces
- Column name is a valid Python identifier

Example:

```python
df.Employee Name    # Error
```

#### 2. **Select Multiple Columns**

````python
df[['Name', 'Salary']]
````

##### **Remember**

- Single Brackets [] → Returns a Series

- Double Brackets [[]] → Returns a DataFrame

````python
result1 = df['Name']
result2 = df[['Name', 'Salary']]
print(type(result1))
print(type(result2))
````

#### 3. **Select All Columns**

````python
df
````

````python
df.columns
````


#####  ***loc* and *iloc* in Pandas**

*loc* and *iloc* are the two main indexing methods in Pandas. They are used to select rows, columns, or both.

*************************************************************************************************************
1. ***iloc* (Integer Position-Based Indexing)**

*iloc* selects data using integer positions.

**Syntax:**

```python
df.iloc[row_position, column_position]
```
*************************************************************************************************************
2. ***loc* (Label-Based Indexing)**

*loc* selects data using row labels (index names) and column labels (column names).

**Syntax:**
```python
df.loc[row_label, column_label]
```
*************************************************************************************************************


#### 4. **Select Columns by Position**

- Using *iloc*

**Syntax:**
```python
df.iloc[rows, columns]
```

- : = all rows
- 0:2 = columns 0 and 1


1. **Select a Single Column Using *iloc***

````python
df.iloc[:,0]
````

2. **Select Multiple Columns Using *iloc***

````python
df.iloc[:, [1,3,5]]
````

3. **Select a Range of Columns Using *iloc***

````python
df.iloc[:, 1:4]
````


#### 5. **Select Columns by Label**

- Using *loc*

**Syntax:**

```python
df.loc[rows, columns]
```

1. **Select Columns using *loc***

````python
df.loc[:, ['Name', 'Department']]
````

2. **Select a Range of Columns Using *loc***

*loc* includes both start and end labels.

````python
df.loc[:, 'Name':'Salary']
````

#### 6. **Select Columns Using a List Variable**

````python
columns = ['Name', 'Experience', 'Salary']
df[columns]
````

#### 7. **Select Columns Based on Data Type**

**Numeric Columns**

````python
df.select_dtypes(include='number')
````

**String Columns**

````python
df.select_dtypes(include='object')
````

````python
df.select_dtypes(include='string')
````

**Most Important Methods for Data Analysis**
```python
df["column"]                  # Single column
df[["col1", "col2"]]          # Multiple columns
df.loc[:, ["col1", "col2"]]   # By labels
df.iloc[:, [0, 2]]            # By positions
df.select_dtypes()            # By data type
df.columns                    # View column names
```
These are the column selection methods you'll use most often in pandas data analysis.
