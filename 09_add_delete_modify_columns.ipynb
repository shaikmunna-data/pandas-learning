### **Add & Delete Columns in Pandas**

#### 1. **ADDING COLUMNS**

There are many ways to add columns.

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
````

##### 1. **Add a Column with a Single Value**

Every row gets the same value.

````python
df['Country'] = 'India'
df.head()
````

##### 2. **Add a Column Using Another Column**

````python
df['Bonus'] = df['Salary'] * 0.10
df.head()
````

##### 3. **Add Multiple Columns**

````python
df['Tax'] = df['Salary'] * 0.05
df['Net Salary'] = df['Salary'] - df['Tax']
df.head()
````

##### 4. **Add an Empty Column**

````python
df['Email'] = None
df.head()
````

##### 5. **Renaming a Column**

````python
df.rename(columns={'Net Salary':'net_salary', 'Email':'email'}, inplace=True)
df.head()
````

Renaming all columns:

````python
df.columns = [
    'emp_id',
    'name',
    'age',
    'department',
    'salary',
    'experience',
    'country',
    'bonus',
    'tax',
    'net_salary',
    'email'
]
df.columns
````

Creating a sample dataframe for exploring more options:

````python
emp = pd.DataFrame({
    "emp_id": [101, 102, 103, 104],
    "name": ["Rahul", "Anita", "John", "Priya"],
    "salary": [35000, 42000, 50000, 38000],
    "department": ["HR", "IT", "Finance", "IT"]
})
emp
````

##### 6. **Add a Column from a List**

The list length must equal the number of rows.

If lengths don't match:

**Output:**

*ValueError:Length of values does not match length of index*

````python
emp["age"] = [25, 30, 28, 26]
emp
````

##### 7. **Insert a Column at a Specific Position**

Normally,

```python
emp["city"] = "Hyderabad"
```
adds the column at the end.


If you want a specific position, use insert().

**Syntax:**

```python
df.insert(position, column_name, values)
```

````python
emp.insert(2, 'gender', ['M', 'F', 'M', 'F'])
emp
````

#### 2. **DELETING COLUMNS**

Several methods are available.

##### 1. ***drop()***

Most commonly used.

**Syntax:**

```python
df.drop(columns=["column_name"])
```

````python
new_df = df.drop(columns=['email'])
print(new_df.head())
print(df.head())    # Original DataFrame remains unchanged.
````

Using *inplace=True*

````python
df.drop(columns=['email'], inplace=True)
df.head()   # Original DataFrame is modified.
````

##### **Delete Multiple Columns**

````python
df.drop(columns=['net_salary', 'tax'], inplace=True)
df.head()
````

##### 2. ***del***

Deletes the column permanently.(can only delete one column at a time)

````python
del df['bonus']
df.head()
````

##### 3. ***pop()***

Removes the column and returns it.

````python
country = df.pop('country')
print(country.head()) # The country column is removed from df.
df.head()
````

##### **Common Errors**

1. List length doesn't match rows

```python
emp["age"] = [25, 30]
```

❌ Error because there are 4 rows.

2. Dropping a non-existing column

```python
emp.drop(columns=["city"])
```

❌ Raises a KeyError.


To avoid the error:

```python
emp.drop(columns=["city"], errors="ignore")
```

#### 4. **Other Modifications**
##### 1. ***assign()***

*assign()* returns a new DataFrame with added or updated columns. It does not modify the original DataFrame unless you assign the result back.

**Syntax:**

```python
df.assign(column_name=value)
```

````python
new_df = df.assign(
    bonus = 5000,
    new_salary = df['salary'] + 5000
)
new_df
````

##### 2. ***astype()***

*astype()* changes the data type of one or more columns.


````python
df.dtypes
````

**Convert integer to  string**

````python
df['age'] = df['age'].astype(str)
df.dtypes
````

**Convert multiple columns**

````python
df[['salary', 'experience', 'age']] = df[['salary', 'experience', 'age']].astype({
    'salary': 'float',
    'experience': 'float',
    'age': int
}
)

print(df.head())
df.dtypes
````
