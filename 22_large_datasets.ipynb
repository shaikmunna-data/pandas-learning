### **Working with Large Datasets in Pandas**

#### 1. **chunksize**

What is it?

Instead of loading the entire file into memory, Pandas loads a fixed number of rows at a time (called a chunk).

This is useful when working with very large CSV files.


**Syntax:**
```python
pd.read_csv(
    "file.csv",
    chunksize=1000
)
```

**Parameter:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Meaning</th>
</tr>
<tr>
<td><i>chunksize</i></td>
<td>Number of rows to read in each chunk</td>
</tr>
</table>

````python
import pandas as pd
df = pd.read_csv(
    r'sample_datasets\large_datasets\movies_dataset.csv',
    usecols=['MovieID', 'Title', 'Genre','Country', 'ReleaseYear','IMDbRating','Director', 'LeadActor']
)
print(df.shape)
print(df.info())
````

##### **Read in chunks**

````python

chunks = pd.read_csv(
    r'sample_datasets\large_datasets\movies_dataset.csv',
    usecols=['MovieID', 'Title', 'Genre','Country','IMDbRating','Director'],
    chunksize=20000
)
for chunk in chunks:
    print(chunk.head)
````

##### **Important**

Iterator can only be used once.

**This works**

```python
chunks = pd.read_csv("employees.csv", chunksize=1000)

for chunk in chunks:
    print(chunk.head())
```

T**his does not work**

```python
chunks = pd.read_csv("employees.csv", chunksize=1000)

for chunk in chunks:
    print(chunk.head())

for data in chunks:
    print(data.head())   # No output
```

Why?

Because chunks is an iterator. Once it reaches the end of the file, it is exhausted.

Read the file again if you need another loop.

#### 2. **iterator**
*iterator=True* creates an iterator object instead of loading all data immediately.

You can manually read chunks using get_chunk().

**Syntax:**
```python
pd.read_csv(
    "file.csv",
    iterator=True
)
```

````python
reader = pd.read_csv(
    r'sample_datasets\large_datasets\movies_dataset.csv',
    usecols=['MovieID', 'Title', 'Genre','Country','IMDbRating','Director'],
    iterator=True
)
````

Read First 10 Rows

````python
reader.get_chunk(10)
````

Read Next 20 Rows

````python
reader.get_chunk(20)
````

Think of a water tap.

**chunksize**

Someone pours one bucket every minute.

**iterator**

You open the tap whenever you want.

#### 3. **Memory Optimization**

Large datasets consume lots of RAM.

You can reduce memory usage by choosing better data types.

````python
df = pd.DataFrame({
    "Age":[25,30,40],
    "Salary":[50000,60000,70000]
})
````

````python
df.info(memory_usage='deep')
````

**Converting DataTypes:**

- Convert Age DataType From int64 to int8
- Convert Salary DataType From int64 to float32

````python
df['Age'] = df['Age'].astype('int8')
df['Salary'] = df['Salary'].astype('float32')
````

**Checking memory_usage Again**

````python
df.info(memory_usage='deep')
````

#### 4. **Categorical Data**

What is categorical data?

Columns that contain a small number of repeated values.

Examples:

- Gender
- Department
- City
- Product Category
- Status

Instead of storing the full text repeatedly, Pandas stores each unique value once and uses integer codes internally.

````python
df = pd.read_csv(
    r'sample_datasets\large_datasets\movies_dataset.csv',
    usecols=['MovieID', 'Title', 'Genre','Country','IMDbRating','Director']
)
````

**Convert to category**

````python
df['Genre'] = df['Genre'].astype('category')
````

**Check datatype**

````python
df.dtypes
````

**Categories**

````python
df['Genre'].cat.categories
````

**Category codes**

````python
df['Genre'].cat.codes
````
