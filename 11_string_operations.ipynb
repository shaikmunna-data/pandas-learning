### **String Operations in Pandas**

String operations are used to work with text data in a DataFrame or Series.

To use string methods in Pandas, use the *.str* accessor.

````python
import pandas as pd

df = pd.DataFrame({
    "emp_id": [101, 102, 103, 104, 105],
    "name": [
        "  alice johnson  ",
        "BOB SMITH",
        "charlie Brown",
        "David Kumar",
        None
    ],
    "email": [
        "alice@gmail.com",
        "BOB@yahoo.com",
        "charlie@gmail.com",
        "david@hotmail.com",
        "emma@gmail.com"
    ],
    "city": [
        "new york",
        "LONDON",
        "mumbai",
        "Hyderabad",
        "CHENNAI"
    ],
    "phone": [
        "9876543210",
        "8765432109",
        "7654321098",
        "6543210987",
        "5432109876"
    ]
})

df
````

#### 1. **Case Conversion**

- *upper()*
- *lower()*
- *capitalize()*
- *title()*

````python

print(df['name'].str.upper())
print(df['name'].str.lower())
print(df['name'].str.capitalize())
print(df['city'].str.title())
````

#### 2. **Remove Spaces**

- *lstrip()*
- *rstrip()*
- *strip()*

````python
print(df['name'].str.lstrip())
print(df['name'].str.rstrip())
print(df['name'].str.strip())
````

#### 3. **Length of String**

- *len()*

````python
df['name'].str.len() - df['name'].str.strip().str.len()
````

#### 4. **Replace Text**

- *replace()*

````python
df['email'].str.replace('gmail', 'outlook').str.replace('yahoo', 'outlook')
````

#### 5. **Check Contains**

- *contains()*

````python
print(df['email'])
df['email'].str.contains('gmail')
````

#### 6. **Startswith / Endswith**

- *startswith()*
- *endswith()*

````python
print(df['email'])
print(df['email'].str.startswith('a'))
print(df['email'].str.endswith('.com'))
````

#### 7. **Split Strings**

- *split()*

````python
df['email'].str.split('@')
````

#### 8. **Extract email domain**

````python
df['email'].str.split('@').str[1]
````

#### 9. **Extract first name**

````python
df['email'].str.split('@').str[0]
````

#### 10. **Get Specific Characters**

````python
print(df['name'])
# Indexing
print(df['name'].str.strip().str[0]) # first character
print(df['name'].str.strip().str[-1]) # last character
# Slicing
print(df['name'].str.strip().str[:3])
````

#### 11. **Count Occurrences**

````python
print(df['email'])
df['email'].str.count('a')
````

#### 12. **Find Position**

````python
print(df['email'])
df['email'].str.index('@')
````

#### 13. **Concatenate Strings**

````python
df['name'].str.strip().str.capitalize()+ '-'+df['city'].str.title()
````

**Real-World Example**


````python
# Clean employees names.
df['name'] = df['name'].str.strip().str.title()
# Emails are usually stored in lowercase.
df['email'] = df['email'].str.strip().str.lower()
# Cities are usually stored in title case.
df['city'] = df['city'].str.strip().str.title()
df # Clean dataframe
````

Instead of writing code repeatedly:

````python
cols = ['name', 'city']
for col in cols:
    df[col] = df[col].str.strip().str.title()

df['city'] = df['city'].str.strip().str.title()

df
````
