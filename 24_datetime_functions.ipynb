### **Date & Time Functions**

Date & time operations are one of the most important pandas topics for data analysis (sales, logs, stock data, attendance, etc.).

#### **Create example dataset**

````python
import pandas as pd

data = {
    'Order_ID': [1001, 1002, 1003, 1004, 1005, 1006],
    'Customer': ['Ravi', 'Asha', 'Kiran', 'Neha', 'Arjun', 'Priya'],
    'City': ['Hyderabad', 'Bengaluru', 'Chennai', 'Mumbai', 'Hyderabad', 'Pune'],
    'Order_Time': [
        '2026-07-20 09:15:00',
        '2026-07-20 14:30:00',
        '2026-07-21 11:45:00',
        '2026-07-21 18:10:00',
        '2026-07-22 08:50:00',
        '2026-07-23 16:20:00'
    ],
    'Delivery_Time': [
        '2026-07-21 12:00:00',
        '2026-07-22 10:00:00',
        '2026-07-22 15:30:00',
        '2026-07-24 09:15:00',
        '2026-07-23 11:20:00',
        '2026-07-25 13:45:00'
    ],
    'Amount': [2500, 1800, 3200, 1500, 4100, 2750]
}

df = pd.DataFrame(data)
df
````

````python
df.dtypes
````

#### 1. **Convert string to datetime**

This is the first thing you should always do.

````python
df['Order_Time'] = pd.to_datetime(df['Order_Time'])
df['Delivery_Time'] = pd.to_datetime(df['Delivery_Time'])
````

````python
df.dtypes
````

#### 2. **Extract date parts (.dt accessor)**

Year, month, day

````python
df['Ord_Year'] = df['Order_Time'].dt.year
df['Ord_Month'] = df['Order_Time'].dt.month
df['Ord_Day'] = df['Order_Time'].dt.day
df[['Order_Time', 'Ord_Year', 'Ord_Month', 'Ord_Day']]
````

#### 3. **Extract Month & Weekday Name**

````python
df['Ord_Month_Name'] = df['Order_Time'].dt.month_name()
df['Ord_Day_Name'] = df['Order_Time'].dt.day_name()
df[['Order_Time', 'Ord_Year', 'Ord_Month_Name', 'Ord_Day_Name']]
````

#### 4. **Extract Time parts**

````python
df['Ord_Time'] = df['Order_Time'].dt.time
df['Ord_Hour'] = df['Order_Time'].dt.hour
df['Ord_Min'] = df['Order_Time'].dt.minute
df[['Order_Time', 'Ord_Time','Ord_Hour', 'Ord_Min']]
````

#### 5. **Extract Only the date**

````python
df['Ord_Date'] = df['Order_Time'].dt.date
df[['Order_Time', 'Ord_Date']]
````

#### 6. **Current date and time**

**Current date&time**

````python
print(pd.Timestamp.now())
````

**Only current data**

````python
print(pd.Timestamp.today().date())
````

#### 7. **Date arithmetic pd.Timedelta() (add/subtract days)**

**Add expected delivery (48 hours)**

````python
df['Expected_Delivery'] = df['Order_Time'] + pd.Timedelta(hours=48)
df[['Order_Time', 'Expected_Delivery']]
````

**Add 7 days**

````python
df['Follow_Up_Date'] = df['Order_Time'] + pd.Timedelta(days=7)
df[['Order_Time', 'Follow_Up_Date']]
````

**Subtract 30 Minutes**

````python
df['Prep_Start'] = df['Order_Time'] - pd.Timedelta(minutes=30)
df[['Order_Time', 'Prep_Start']]
````

#### 8. **Difference between two datetime columns**

Calculate delivery duration

````python
df['Delivery_Duration'] = df['Delivery_Time'] - df['Order_Time']
df[['Order_Time', 'Delivery_Time', 'Delivery_Duration']]
````

#### 9. **Filter by date**

Orders placed on July 21

````python
df.loc[df['Order_Time'].dt.date == pd.to_datetime('2026-07-21').date(), 'Order_ID':'Amount']
````

Orders after July 21

````python
df.loc[df['Order_Time'] > '2026-07-21', 'Order_ID':'Amount']
````

#### 10. **Set datetime as index**

````python
df_ts = df.set_index('Order_Time') #Rule: Keep datetime64[ns] for pandas time-series operations; .dt.date converts it to object dtype.
df_ts.head()
````

#### 11. **Date range selection**

Orders from July 20 to July 21

````python
df_ts.loc['2026-07-20':'2026-07-21', 'Order_ID':'Amount']
````

Only July 20

````python
df_ts.loc['2026-07-20', 'Order_ID':'Amount']
````

#### 12. **strftime() — Format for reports**

Convert to Indian report format

````python
df['Report_Date'] = df['Order_Time'].dt.strftime('%d-%m-%Y %I:%M %p')
print(df[['Order_Time', 'Report_Date']])
````

##### **Best Practice**

Always use one of these:

```python
# If the input date is already in ISO format (YYYY-MM-DD)
pd.to_datetime("2026-07-24")

# If the input date is in another format, specify it explicitly
pd.to_datetime(df["Date"], format="%d-%m-%Y")
```

**Note:**

- `format=` does not change the date format.
- It tells pandas how to read (parse) the input date.
- The `format` must match the input date format.

**Memory Trick:**

- `pd.to_datetime(format=...)` → Read the date.
- `.dt.strftime(...)` → Display the date in a different format.
