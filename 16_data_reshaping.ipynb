### **Reshaping Data**

Reshaping Data means changing the structure or layout of a DataFrame without changing the actual data. It helps you organize data into a format that's easier to analyze or visualize.

Some of the most commonly used reshaping methods in Pandas include:

- *pivot()*
- *pivot_table()*
- *melt()*
- *stack()*
- *unstack()*
- *explode()*
- *transpose() (.T)*
- *set_index()*
- *reset_index()*
- *wide_to_long()*

#### 1. **Pivot()**

pivot() reshapes a DataFrame by turning unique values form one column into new columns.

- Convert rows into columns.
- Does not perform aggregation.
- Requires unique values.

**Syntax:**
```python
DataFrame.pivot( 
    data=df,
    index=None, 
    columns=None, 
    values=None
)
```

**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
</tr>
<tr>
<td><i>data</i></td>
<td>DataFrame to summarize</td>
</tr>
<tr>
<td><i>index</i></td>
<td>Column to use as the row index.</td>
</tr>
<tr>
<td><i>columns</i></td>
<td>Column whose unique values become new column names.</td>
</tr>
<tr>
<td><i>values</i></td>
<td>Column whose values fill the new table.</td>
</tr>
</table>

````python
import pandas as pd

df = pd.DataFrame({
    "Date": ["2025-01-01", "2025-01-01", "2025-01-02", "2025-01-02"],
    "Product": ["Laptop", "Phone", "Laptop", "Phone"],
    "Sales": [50000, 20000, 60000, 25000]
})

print(df)
````

````python

pivot_1 = pd.pivot(
    df,
    index='Date',
    columns='Product',
    values='Sales'

)
pivot_1
````

#### 2. **Pivot Tables**
A pivot table summarizes and aggregates data, similar to Excel's PivotTable.

- Similar to *pivot()*, but supports aggregation.
- Can handle duplicate values.

**Syntax:**
```python
pd.pivot_table(
    data=df,              # DataFrame to summarize
    values='Sales',       # Column to aggregate
    index='Region',       # Row groups
    columns='Category',   # Column groups
    aggfunc='sum',        # Aggregate using sum
    fill_value=0,         # Replace NaN with 0
    margins=True          # Add grand totals
)

````python
sales = pd.read_csv(r'sample_datasets\Sales.csv')
sales
````

````python
pivot_table1 = pd.pivot_table(
    sales,
    index='Year',
    columns= ['Category', 'Product'],
    values='Sales',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='Total'
)
pivot_table1
````

#### 3. **melt()**

melt() converts multiple columns into rows, producing a long-format DataFrame

*pandas.melt()* is used to unpivot a DataFrame from a wide format to a long format.

**Syntax:**
```python
pd.melt(
    frame,
    id_vars=None,
    value_vars=None,
    var_name=None,
    value_name='value'
)
```
**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>What it does</th>
<th>Default</th>
</tr>
<tr>
<td><i>frame</i></td>
<td>The DataFrame to reshape</td>
<td>Required</td>
</tr>
<tr>
<td><i>id_vars</i></td>
<td>Columns to keep unchanged (identifiers)</td>
<td>None</td>
</tr>
<tr>
<td><i>value_vars</i></td>
<td>Columns to convert into rows</td>
<td>All columns except id_vars</td>
</tr>
<tr>
<td><i>var_name</i></td>
<td>Name of the new column holding original column names</td>
<td>"variable"</td>
</tr>
<tr>
<td><i>value_name</i></td>
<td>Name of the new column holding values</td>
<td>"value"</td>
</tr>
<tr>
<td><i>ignore_index</i></td>
<td>Whether to reset the index (True) or keep the original index (False)</td>
<td>True</td>
</tr>
</table>

````python
students = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank'],
    'Class': ['10A', '10A', '10B', '10B', '10A', '10B'],
    'Math': [85, 78, 92, 88, 95, 81],
    'Science': [90, 82, 89, 91, 96, 85],
    'English': [88, 80, 94, 86, 92, 83],
    'Computer': [95, 84, 97, 90, 99, 88]
})

students
````

**Example 1: Keep only Name as identifier**

Since only Name is an identifier, every other column is melted, including:

- Student_ID
- Class
- Math
- Science
- English
- Computer

````python
unpivot_stu1 = pd.melt(
    students,
    id_vars='Name'
)
unpivot_stu1
````

**Example 2: Melt only Math and Science**

````python
unpivot_stu2 = pd.melt(
    students,
    id_vars=['Student_ID', 'Name'],
    value_vars=['Math', 'Science'],
    var_name='Subject',
    value_name='Marks'
)
unpivot_stu2
````

**Example 3: Melt all subject columns**

````python
unpivot_stu3 = pd.melt(
    students,
    id_vars=['Student_ID', 'Name', 'Class'],
    var_name='Subject',
    value_name='Marks'
)
    # you didn't specify value_vars, so pandas uses its default behavior.
    #If value_vars=None (the default), pandas automatically selects all columns that are NOT in id_vars
unpivot_stu3
````

#### 4. **Stack()**

*stack()* is a reshaping method in pandas that moves one or more column levels into the row index.

**Syntax:**
```python
DataFrame.stack(level=-1, future_stack=True)
```
**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Default</th>
</tr>
<tr>
<td><i>level</i></td>
<td>Column level(s) to stack (for MultiIndex columns)</td>
<td>-1</td>
</tr>
<tr>
<td><i>future_stack</i></td>
<td>Uses the newer implementation</td>
<td>True</td>
</tr>
</table>


````python
df = pd.DataFrame({
    'Math': [90, 80, 70, 85],
    'Science': [85, 88, 73, 80]
},
index=['Alice', 'Bob', 'Charlie', 'David'])

print(df)
df.index
````

````python
stacked1 = df.stack()
print(stacked1)
print(stacked1.index),
print(type(stacked1))
````

**Example with Missing Values**

````python
df = pd.DataFrame({
    'Math': [90, None, 70, 85],
    'Science': [85, 88, None, 80]
},
index=['Alice', 'Bob', 'Charlie', 'David'])
df
````

Stack (By Default Stack keeps missing values)

````python
stacked2 = df.stack()
stacked2
````

Removing missng values

````python
stacked3 = df.stack().dropna()
stacked3
````

**Example with MultiIndex Columns**

````python
arrays = [
    ['Marks', 'Marks', 'Grade', 'Grade'],
    ['Math', 'Science', 'Math', 'Science']
]

columns = pd.MultiIndex.from_arrays(arrays)
df = pd.DataFrame([
    [90, 85, 'A', 'B'],
    [80, 88, 'B', 'A']
], index=['Alice', 'Bob'], columns=columns)
df
````

**The columns now have 2 levels.**

**Level 0**
- Marks
- Marks
- Grade
- Grade

**Level 1 or (-1)**
- Math
- Science
- Math
- Science

**Stack the last column level (default)**

````python
print(df.stack())
stacked4 = df.stack(level=-1)
stacked4
````

**Stack column level 0**

````python
stacked5 = df.stack(level=0)
stacked5
````

**Stack both column levels**

````python
stacked6 = df.stack(level=[0,1])
stacked6
````

#### 4. **unstack()**

*unstack()* does the opposite to *stack()*.

It converts an index level into columns.

**Example 1 — Basic unstack()**

````python
print(stacked6)
unstacked1 = stacked6.unstack()
unstacked1
````

**Unstack by level number**

From this, you can tell:


Level 0

- Student (Alice, Bob)

Level 1

- Type (Marks, Grade)

Level 2 

-Subject (Math, Science)

**Move the last level (Subject)**

````python
print(stacked6)
unstacked2 = stacked6.unstack(-1)
unstacked2
````

**Move the second level (Type)**

````python
unstacked3 = stacked6.unstack(level=1)
unstacked3
````

**Move the first level (Student)**

````python
unstacked4 = stacked6.unstack(level=0)
unstacked4
````

**Move last two levels(Subject/Type)**

````python
unstacked5 = stacked6.unstack(level=[1,2])
unstacked5
````

#### 5. **explode()**

*explode()* is a Pandas method used to split list-like values in a column into multiple rows.

It is very useful when a single cell contains multiple values (like a list) and you want one value per row.


**Syntax:**
```python
df.explode(column, ignore_index=False)
```

**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
</tr>
<tr>
<td><i>column</i></td>
<td>Column name (or list of column names) to explode.</td>
</tr>
<tr>
<td><i>ignore_index</i></td>
<td>If True, creates a new index (0,1,2...). Default is False.</td>
</tr>
</table>

**Sample Datasets to practice *explode()***

````python
students1 = pd.DataFrame({
    'Student': ['Alice', 'Bob', 'Charlie'],
    'Subjects': [
        ['Math', 'Science'],
        ['English'],
        ['Physics', 'Chemistry', 'Biology']
    ]
})

orders = pd.DataFrame({
    'Order_ID': [101, 102, 103],
    'Products': [
        ['Laptop', 'Mouse'],
        ['Keyboard'],
        ['Monitor', 'Webcam', 'Speaker']
    ]
})

post = pd.DataFrame({
    'Post': ['P1', 'P2'],
    'Tags': [
        ['Python', 'Pandas', 'Data'],
        ['SQL', 'MySQL']
    ]
})
print(students1)
print()
print(orders)
print()
print(post)
````

**Exploding One Column**

````python
print(post)
post.explode('Tags')
````

````python
print(orders)
orders.explode('Products')
````

````python
print(students1)
students1.explode('Subjects')
````

Notice:

- Alice appears twice.
- Charlie appears three times.
- The index is repeated.

**Reset the Index**

````python
students1.explode('Subjects', ignore_index=True)
````

````python
orders.explode('Products', ignore_index=True)
````

````python
post.explode('Tags', ignore_index=True)
````

**Exploding Multiple Columns**

If two columns contain lists of the same length, you can explode them together.

````python
students2 = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Subject': [
        ['Math', 'Science'],
        ['English', 'History']
    ],
    'Marks': [
        [90, 85],
        [88, 92]
    ]
})
students2
````

````python
students2.explode(['Subject', 'Marks'], ignore_index=True)
````

> **Important:** When exploding multiple columns, the lists in each row must have the same length.
