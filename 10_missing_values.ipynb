### **Missing Values in Pandas**

Missing values are values that are not available or unknown in a dataset.

In Pandas, missing values are usually represented as:

- NaN (Not a Number) – for numeric data
- None – for object/string data (often treated as missing)
- pd.NA – the newer missing value marker used by Pandas

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees_with_Nan.csv')
df
````

#### 1. **Check for Missing Values**

***isna()*** or ***isnull()***

Returns *True* where a value is missing

````python
#df.isna() 
df.isnull()
````

#### 2. **Check Non-Missing Values**

***notna()*** or ***notnull()***

````python
df.notnull() # or df.notna()
````

#### 3. **Count Missing Values**

````python
df.isnull().sum() 
````

Total missing values

````python
df.isnull().sum().sum()
````

#### 4. **Check if Any Missing Values Exist**

````python
df.isnull().any()
````

#### 5. **Check if All Values Are Missing**

````python
df.isnull().all()
````

#### 6. **Retrieve Rows with Missing Data**

It returns all rows where at least one column contains a missing value.

````python
df[
    (df['Name'].isnull()) | 
    (df['Age'].isnull()) | 
    (df['Department'].isnull()) | 
    (df['Salary'].isnull()) | 
    (df['Experience'].isnull())
    ] # It returns all rows where at least one column contains a missing value.
````

#### **A much shorter and better way**

Instead of checking every column manually

```python
df[df.isna().any(axis=1)] 
#or
df[df.isnull().any(axis=1)]
```

````python
df[df.isnull().any(axis=1)]
````

#### 7. **Removing Missing Values *dropna()***

- Remove rows with missing values.

````python
print(df.dropna())
df.dropna().isnull().sum()
````

- Remove rows only if all values are missing

````python
print(df.dropna(how='all'))
df.dropna(how='all').isnull().sum()
````

- Remove rows if any value is missing

````python
print(df.dropna(how='any'))
df.dropna(how='any').isnull().sum()
````

- Keep rows with at least 5 non-missing values

````python
print(df.dropna(thresh=5))
df.dropna(thresh=5).isnull().sum()
````

#### 8. **Filling Missing Values**

- Fill with a constant

````python
df.fillna(0)
````

- Fill a single column

````python
df['Age'] = df['Age'].fillna(20)
df
````

- Fill different columns with different values

````python

df.fillna({
    'Name': 'Unknown',
    'Age': 20,
    'Department': 'Unknown',
    'Salary': 'N/A',
    'Experience': 'N/A'
}, inplace=True)
df
````

- **Forward Fill (ffill())**

Rule: Fill missing values using the value from the previous row.

````python


new_df = pd.DataFrame({
    "Name": ["Alice", None, "Charlie", None, "Eve"],
    "Salary": [50000, None, 60000, None, 70000]
})

new_df
````

````python
new_df.ffill()
````

- **Backward Fill (bfill())**

Rule: Fill missing values using the value from the next row.

````python
new_df.bfill()
````

- **Replace Missing Values(replace())**

````python
new_df.replace('Alice', 'Bob')
````

````python
df.replace('Finance', 'Finance & Accounting').tail(3)
````
