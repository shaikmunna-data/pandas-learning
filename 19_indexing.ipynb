#### 1. **set_index()**

*set_index()* changes one or more columns into the DataFrame's index.


**Syntax:**
```python
df.set_index(keys, drop=True, inplace=False)
```
**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
</tr>
<tr>
<td><i>keys</i></td>
<td>Column(s) to become index</td>
</tr>
<tr>
<td><i>drop=True</i></td>
<td>Remove column after making it index</td>
</tr>
<tr>
<td><i>drop=False</i></td>
<td>Keep original column</td>
</tr>
<tr>
<td><i>inplace=True</i></td>
<td>Modify original DataFrame</td>
</tr>
</table>

````python
import pandas as pd
# Sample data
df = pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Name':['Alice','Bob','Charlie','David'],
    'Math':[90,75,88,95]
})

print(df)
````

##### **Example 1 (Single Index)**

````python
df.set_index('Student_ID', inplace=True)
df
````

#### 2. **reset_index()**

Converts the index back into normal columns.

**Syntax:**
```python
df.reset_index(drop=False, inplace=False)
```

**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
</tr>
<tr>
<td><i>drop</i></td>
<td>If True, discard the index instead of creating a column</td>
</tr>
<tr>
<td><i>inplace</i></td>
<td>Modify original DataFrame</td>
</tr>
</table>

````python
print(df)
df.reset_index(inplace=True)
df
````

#### 3. **MultiIndex**


A MultiIndex is an index with multiple levels.

````python
df = pd.DataFrame({
    'Region':['East','East','West','West'],
    'Year':[2024,2025,2024,2025],
    'Sales':[1000,1200,900,1100]
})

df = df.set_index(['Region','Year'])

print(df)
````

**Access data**

- Entire Region

````python
df.loc['West']
````

- Specific Row

````python
df.loc[('East', 2024)]
````
