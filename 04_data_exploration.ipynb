### **Data Exploration**

Data exploration means understanding your dataset before analyzing or processing it. It helps you answer questions like:

- How many rows and columns are there?
- What are the column names?
- What are the data types?
- Are there missing values?
- What are the basic statistics?

```python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
```

#### 1. **head()**

Displays the first 5 rows by default.

```python
df.head()
```

```python
df.head(3) # Displays first 3 rows
```

#### 2. **tail()**

Displays the last 5 rows.

```python
df.tail()
```

```python
df.tail(3) # Displays the last 3 rows
```

#### 3. **shape**

Returns the number of rows and columns.

```python
df.shape
```

Meaning:
- 100 rows
- 6 columns

#### 4. **columns**

Shows all column names.

```python
df.columns
```

Convert to a list:

```python
list(df.columns)
```

#### 5. **index**

Shows row labels.

```python
df.index
```

#### 7. **info()**

Provides a summary of the DataFrame.

It shows:

- Number of rows
- Number of columns
- Non-null values
- Data types
- Memory usage

```python
df.info()
```

#### 8. **describe()**

Shows statistical summary for numeric columns.

```python
df.describe()
```

Meaning:

- *count* → Number of values
- *mean* → Average
- *std* → Standard deviation
- *min* → Smallest value
- *25%, 50%, 75%* → Quartiles
- *max* → Largest value

#### 9. **isnull() or isna()**

Checks for missing values.

Returns: *True/False*

```python
df.isnull()
```

Count missing values:

```python
df.isnull().sum()
```

#### 10. **nunique()**

Counts unique values.

```python
df.nunique()
```

#### 11. **value_counts()**

Counts occurrences of unique values.

```python
df['Department'].value_counts()
```

```python
df['Name'].value_counts()
```

#### 12. **sample()**

Returns random rows.

```python
df.sample(4)
```

```python
df.sample(3)
```
