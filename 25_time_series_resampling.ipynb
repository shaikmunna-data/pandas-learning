### **Resampling**

**Resampling means:**

Changing the frequency of time-series data (daily → monthly, monthly → yearly, hourly → daily, etc.) and then applying an aggregation function.

It only works with dates

#### **There are two common ways to use resample().**
----

1. **When the Date is the Index (Most Common)**

**Syntax:**
```python
df.resample(rule)[column].aggregation()
```
---
2. **When the Date is NOT the Index**

**Syntax:**
```python
df.resample(rule, on="date_column")[column].aggregation()
```
---

````python
import pandas as pd

df = pd.DataFrame({
    "Order_Date":[
        "2025-01-03",
        "2025-01-10",
        "2025-01-15",
        "2025-02-05",
        "2025-02-18",
        "2025-03-12",
        "2025-03-25",
        "2025-04-08",
        "2025-04-22",
        "2025-05-05",
        "2025-06-10",
        "2025-07-15",
        "2025-08-20",
        "2025-09-10",
        "2025-10-18",
        "2025-11-25",
        "2025-12-30",
        "2026-01-12",
        "2026-02-20",
        "2026-03-28"
    ],

    "Sales":[
        1200,1800,1400,
        2000,1700,
        2400,2600,
        1900,2200,
        2500,
        2700,
        2900,
        3100,
        2800,
        3300,
        3500,
        4000,
        4200,
        4500,
        4800
    ],

    "Profit":[
        200,350,250,
        420,360,
        500,550,
        410,450,
        520,
        600,
        650,
        700,
        620,
        760,
        800,
        950,
        1000,
        1050,
        1100
    ],

    "Orders":[
        15,18,16,
        21,19,
        24,26,
        20,22,
        25,
        28,
        30,
        33,
        29,
        35,
        37,
        40,
        42,
        45,
        48
    ]
})
````

**Convert the date column.**

````python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
print(df.dtypes)
df
````

#### **Method 1: Resampling using Date Index (Most Common)**

Set the date column as the index.

````python
df_ts = df.set_index('Order_Date')
print(df_ts.index)
````

**Monthly Sales**

````python
df_ts.resample('ME')['Sales'].sum()
````

**Monthly Profit**

````python
df_ts.resample('ME')['Profit'].sum()
````

**Monthly Average Sales**

````python
df_ts.resample('ME')['Sales'].mean()
````

**Yearly Resampling**

````python
df_ts.resample('YE').sum()
````

**Multiple Aggregations**

````python
df_ts.resample('YE').agg({
    'Sales':['sum', 'mean', 'max'],
    'Profit':['sum', 'mean'],
    'Orders':['sum']
})
````

#### **Method 2: Resampling WITHOUT Setting the Index**

Sometimes you don't want to change the index.

Use the on parameter.

**Monthly Sales**

````python
df.resample('ME', on='Order_Date')['Sales'].sum()
````

**Monthly Profit**

````python
df.resample('ME', on='Order_Date')['Profit'].sum()
````

**Yearly**

````python
df.resample('YE', on='Order_Date').sum()
````

##### **When to Use Which?**

**Use groupby() when your groups are based on categories:**

- Department
- Product
- City
- Customer
- Employee

**Use resample() when your groups are based on time:**

- Daily sales
- Weekly website visitors
- Monthly revenue
- Quarterly profit
- Yearly expenses
