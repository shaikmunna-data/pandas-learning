### **Window Functions in Pandas**

A window function performs calculations over a group of rows related to the current row without collapsing the DataFrame.

#### **Types of Window Functions in Pandas**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Window Function</th>
<th>Purpose</th>
<th>Industry Use</th>
</tr>
<tr>
<td><i>Rolling</i></td>
<td>Fixed-size moving window</td>
<td>Stock analysis, moving average</td>
</tr>
<tr>
<td><i>Expanding</i></td>
<td>From first row to current row</td>
<td>Running totals</td>
</tr>
<tr>
<td><i>Exponential Moving (EWM)</i></td>
<td>Gives more importance to recent data</td>
<td>Forecasting</td>
</tr>
<tr>
<td><i>Shift</i></td>
<td>Move data up/down</td>
<td>Previous day comparison</td>
</tr>
<tr>
<td><i>Diff</i></td>
<td>Difference between rows</td>
<td>Daily change</td>
</tr>
<tr>
<td><i>Rank</i></td>
<td>Ranking within dataset/group</td>
<td>Sales ranking</td>
</tr>
</table>

These are the six primary window-style operations most commonly used in pandas.

````python
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range("2025-01-01", periods=15),
    "Sales":[
        1200,1500,1700,1600,2000,
        1800,2200,2500,2400,2600,
        2800,3000,3100,3300,3500
    ],
    "Orders":[
        22,25,28,27,31,
        30,35,37,36,39,
        40,42,43,45,47
    ],
    "Visitors":[
        210,250,270,260,300,
        295,330,360,355,380,
        400,420,430,450,470
    ]
})

df
````

#### 1. **Rolling Window**

**What it does**

Looks at a fixed number of consecutive rows.

**Example:**

**Window = 3**

Current row uses
```
Row1
Row2
Row3
```
Next row

```
Row2
Row3
Row4
```

**Syntax:**
```python
df["column"].rolling(
    window,
    min_periods=None,
    center=False,
    win_type=None,
    on=None,
    closed=None
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Example</th>
</tr>
<tr>
<td><i>window</i></td>
<td>Size of the moving window</td>
<td>window=3</td>
</tr>
<tr>
<td><i>min_periods</i></td>
<td>Minimum observations required to calculate a value</td>
<td>min_periods=1</td>
</tr>
<tr>
<td><i>center</i></td>
<td>Place result at center of window instead of right edge</td>
<td>center=True</td>
</tr>
<tr>
<td><i>win_type</i></td>
<td>Weighted window type (e.g., Gaussian, triangular)</td>
<td>win_type="gaussian"</td>
</tr>
<tr>
<td><i>on</i></td>
<td>Column to use for rolling when DataFrame index isn't datetime</td>
<td>on="Date"</td>
</tr>
<tr>
<td><i>closed</i></td>
<td>Which window endpoints are included (time-based windows)</td>
<td>closed="right"</td>
</tr>
</table>

 **USE CASES**
 
1. **Moving Average**

````python
df['Sales_MA_3'] = df['Sales'].rolling(window=3).mean() #Today's average sales over the last 3 days.
df
````

2. **Rolling Sum**

````python
df['Sales_Last3Days'] = df['Sales'].rolling(window=3).sum() #How much revenue came in during the last 3 days?
df
````

3. **Rolling Maximum**

````python
df['Highest_Sales'] = df['Sales'].rolling(window=7).max() #Highest sales achieved during the past week.
df
````

4. **Rolling Minimum**

````python
df['Lowest_Sales'] = df['Sales'].rolling(window=7).min() #Worst sales day during the past week.
df
````

#### 2. **Expanding Window**

**What it does**

Starts from the first row and keeps growing.

```
Row1

↓

Row1 Row2

↓

Row1 Row2 Row3

↓

Row1 Row2 Row3 Row4
```

**Syntax:**
```python
df["column"].expanding(
    min_periods=1,
    axis=0
)
```


**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Example</th>
</tr>
<tr>
<td><i>min_periods</i></td>
<td>Minimum rows before calculation begins</td>
<td>min_periods=2</td>
</tr>
<tr>
<td><i>axis</i></td>
<td>Compute across rows or columns (axis is deprecated for some APIs; rows is the common default)</td>
<td>axis=0</td>
</tr>
</table>


**USE CASES**

1. **Running Total**

````python
df2 = pd.DataFrame({
    "Date": pd.date_range("2025-01-01", periods=15),
    "Sales":[
        1200,1500,1700,1600,2000,
        1800,2200,2500,2400,2600,
        2800,3000,3100,3300,3500
    ],
    "Orders":[
        22,25,28,27,31,
        30,35,37,36,39,
        40,42,43,45,47
    ],
    "Visitors":[
        210,250,270,260,300,
        295,330,360,355,380,
        400,420,430,450,470
    ]
})
````

````python
df2['Running_Sales'] = df2['Sales'].expanding().sum() #Total sales since business started.
df2
````

2. **Running Average**

````python
df2['Average_Sales'] = df2['Sales'].expanding().mean() #Average sales until today.
df2
````

3. **Running Maximum**

````python
df2['Best_Sales_Ever'] = df2['Sales'].expanding().max() #Highest sales ever achieved.
df2
````

4. **Running Minimum**

````python
df2['Lowest_Sales_Ever'] = df2['Sales'].expanding().min()   #Lowest sales ever recorded.
df2
````

#### 3. **Exponentially Weighted Window (EWM)**

**What it does**

Recent observations get more weight than older ones.

Older data never completely disappears, but its influence fades over time.

**Syntax:**
```python
df["column"].ewm(span=3).mean()
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Example</th>
</tr>
<tr>
<td><i>span</i></td>
<td>Most commonly used smoothing parameter</td>
<td>span=5</td>
</tr>
<tr>
<td><i>alpha</i></td>
<td>Smoothing factor (0–1)</td>
<td>alpha=0.3</td>
</tr>
<tr>
<td><i>com</i></td>
<td>Center of mass (alternative to span)</td>
<td>com=2</td>
</tr>
<tr>
<td><i>halflife</i></td>
<td>Weight halves after this period</td>
<td>halflife=4</td>
</tr>
<tr>
<td><i>adjust</i></td>
<td>Whether to normalize weights (default True)</td>
<td>adjust=False</td>
</tr>
<tr>
<td><i>min_periods</i></td>
<td>Minimum observations before output</td>
<td>min_periods=2</td>
</tr>
<tr>
<td><i>ignore_na</i></td>
<td>Ignore missing values when weighting</td>
<td>ignore_na=True</td>
</tr>
</table>

>Important: Use only one of span, alpha, com, or halflife because they all define the smoothing in different ways.


````python
df2["EMA"] = (df2["Sales"].ewm(span=5).mean()) #Forecast sales trends while reacting faster to recent changes.
df2
````

#### 4. **Shift**

**What it does**

Moves values up or down.

No calculations.

**Syntax:**
```python
df["column"].shift(
    periods=1,
    freq=None,
    fill_value=None
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Example</th>
</tr>
<tr>
<td><i>periods</i></td>
<td>Number of rows to shift</td>
<td>periods=1</td>
</tr>
<tr>
<td><i>freq</i></td>
<td>Shift by a calendar frequency (requires datetime index)</td>
<td>freq="D"</td>
</tr>
<tr>
<td><i>fill_value</i></td>
<td>Value used for new missing positions</td>
<td>fill_value=0</td>
</tr>
</table>



**Use Cases**

1. **Previous Day Sales**


````python
df3 = pd.DataFrame({
    "Date": pd.date_range("2025-01-01", periods=15),
    "Sales":[
        1200,1500,1700,1600,2000,
        1800,2200,2500,2400,2600,
        2800,3000,3100,3300,3500
    ],
    "Orders":[
        22,25,28,27,31,
        30,35,37,36,39,
        40,42,43,45,47
    ],
    "Visitors":[
        210,250,270,260,300,
        295,330,360,355,380,
        400,420,430,450,470
    ]
})
````

````python
df3['Yesterday_Sales'] = df3['Sales'].shift()
df3
````

2. **Next Day Sales**

````python
df3['Tomorrow_Sales'] = df3['Sales'].shift(-1)
df3
````

#### 5. **Diff**

**What it does**

Subtracts the previous value from the current value.

**Syntax:**
```python
df["column"].diff(
    periods=1
)
```
**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Example</th>
</tr>
<tr>
<td><i>periods</i></td>
<td>Difference between current row and row <i>n</i> periods away</td>
<td>periods=1</td>
</tr>
</table>

**Use Cases**
**Daily Sales Increase**

````python
df3['Sales_Change'] = (df3['Sales'].diff())
df3
````

#### 6. **Rank**

**What it does**

Assigns rankings based on values.

**Syntax:**
```python
df["column"].rank(
    axis=0,
    method="average",
    numeric_only=False,
    na_option="keep",
    ascending=True,
    pct=False
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
<th>Example</th>
</tr>
<tr>
<td><i>ascending</i></td>
<td>Rank from smallest (True) or largest (False)</td>
<td>ascending=False</td>
</tr>
<tr>
<td><i>method</i></td>
<td>How to rank ties</td>
<td>method="dense"</td>
</tr>
<tr>
<td><i>na_option</i></td>
<td>Handle missing values</td>
<td>na_option="bottom"</td>
</tr>
<tr>
<td><i>pct</i></td>
<td>Return percentile ranks</td>
<td>pct=True</td>
</tr>
<tr>
<td><i>numeric_only</i></td>
<td>Rank only numeric columns (for DataFrames)</td>
<td>numeric_only=True</td>
</tr>
</table>


**Use Cases**

**Top Selling Products**

````python
df4 = pd.DataFrame({
    "Date": pd.date_range("2025-01-01", periods=15),
    "Sales":[
        1200,1500,1700,1600,2000,
        1800,2200,2500,2400,2600,
        2800,3000,3100,3300,3500
    ],
    "Orders":[
        22,25,28,27,31,
        30,35,37,36,39,
        40,42,43,45,47
    ],
    "Visitors":[
        210,250,270,260,300,
        295,330,360,355,380,
        400,420,430,450,470
    ]
})
````

````python
df4['Sales_Rank'] = df4['Sales'].rank(ascending=False)
df4
````
