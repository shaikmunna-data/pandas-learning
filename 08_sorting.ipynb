### **Sorting in Pandas**

Sorting means arranging data in a specific order.

There are two types of sorting:

1. **Sort Rows (sort_values())**
2. **Sort Index (sort_index())**

Let's learn them one by one.

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
````

#### 1. **sort_values()**

Used to sort rows based on column values.

**Syntax:**

```python
df.sort_values(by="column_name")
```

##### **Sort by one column (Ascending)**

Ascending is the default.

````python
df.sort_values(by='Age')
````

##### **Sort by (Descending)**

````python
df.sort_values(by='Age', ascending=False)
````

##### **Sort by Multiple Columns**

Suppose two people have the same age.

Sort by Age first and then Salary.

Explanation

- First sort by Age.
- If Age is the same, sort by Salary.

````python
df.sort_values(by=['Age', 'Salary'])
````

##### **Different Order for Different Columns**

Meaning

- Age → Ascending
- Salary → Descending

````python
df.sort_values(by=['Age', 'Salary'], ascending=[True, False])
````

````python
df2 = pd.read_csv(r'sample_datasets\Employees.csv', na_values=['Engineering'])
df2
````

##### **Handling Missing Values**

By Default:
- Missing values appear at the end.

````python
df2.sort_values(by='Department')
````

Move missing values first

````python
df2.sort_values(by='Department', na_position='first')
````

##### **Ignore Old Index**

It creates a new index (0, 1, 2, ...) after sorting.

**Without *ignore_index=True***

````python
df.sort_values(by='Age').head()
````

**With *ignore_index=True***

````python
df.sort_values(by='Age', ignore_index=True).head()
````

##### **In-place Sorting**

It changes the original DataFrame.

**Without *inplace=True***

- Shows the sorted result.
- df is NOT changed.

````python
df.sort_values(by='Age') # # Original DataFrame remains unchanged.
df 
````

**With *inplace=True***
- Sorts the data.
- df is permanently changed.

````python
df.sort_values(by='Age', inplace=False) # Now df is permanently sorted.
df
````

##### **Case-sensitive Sorting**

Uppercase letters come before lowercase because sorting is based on Unicode values.

````python
df3 = pd.DataFrame({
    "Name":["john","Alice","bob","David"]
})
df3.sort_values(by='Name')
````

To sort without considering case:

````python
df3.sort_values(by='Name', key=lambda col: col.str.lower())
````

##### **Custom Sorting Using *key***

Example: Sort names by their length.

````python
df.sort_values(by='Name', key=lambda col : col.str.len())
````

#### 2. **sort_index()**

Sorts rows or columns by their index labels, not by the data.

##### **Sort Row Index**

````python
df.sort_index(ascending=False)
````

##### **Sort Column Names**

axis=1 means sort column labels.

````python
df.sort_index(axis=1)
````
