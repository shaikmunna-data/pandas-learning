#### 1. ***merge()***

Purpose: Combine DataFrames based on one or more common columns.

**Syntax:**
```python
pd.merge(
    left,
    right,
    how='inner',
    on=None,
    left_on=None,
    right_on=None,
    left_index=False,
    right_index=False,
    suffixes=['_x', '_y']
)
```

**Parameters:**
- *left*: Left DataFrame.
- *right*: Right DataFrame.
- *how='inner'*: Type of join.
    - *'inner'* → Matching rows only.
    - *'left'* → All rows from left DataFrame.
    - *'right'* → All rows from right DataFrame.
    - *'outer'* → All rows from both DataFrames.
    - *'cross'* → Cartesian product (every row with every row).
- *on='column_name'*: Common column present in both DataFrames.
- *left_on='column'*: Join column from the left DataFrame.
- *right_on='column'*: Join column from the right DataFrame.
- *left_index=True*: Use the left DataFrame's index as the key.
- *right_index=True*: Use the right DataFrame's index as the key.
- *suffixes=['_x', '_y']*: Rename duplicate column names after merging.
- *indicator=False:*
    - *False (default)*: Don't add an indicator column.
    - *True*: Add a column named *_merge* showing the source of each row.
    - You can also provide a string (e.g., *indicator='source'*) to use a custom column name.

##### 1. **Inner Join (default)**

Keeps only matching rows.

````python
inner_join = pd.merge(customers, orders, on='customer_id')
inner_join
````

##### 2. **Left Join**

Keeps all rows from the left DataFrame.

````python
left_join_1 = pd.merge(customers, orders, on='customer_id', how='left')
left_join_1
````

````python
left_join_2 = pd.merge(orders, customers, on='customer_id', how='left')
left_join_2
````

##### 3. **Right Join**

Keeps all rows from the right DataFrame.

Many people simply avoid *right joins* because the *left* version is often easier to read.

````python
right_join = pd.merge(customers, orders, on='customer_id', how='right')
right_join
````

##### 4. **Outer Join**

Keeps every row from both DataFrames.

````python
outer_join = pd.merge(customers, orders, on='customer_id', how='outer')
outer_join
````

##### 5. **Merge on Different Column Names**

When to use

When the matching columns have different names in each DataFrame.

````python
# Employee Data
employees = pd.DataFrame({
    "employee_id": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"]
})
# Salary Data
salary = pd.DataFrame({
    "id": [101, 102, 104, 105],
    "Salary": [50000, 60000, 70000, 80000]
})
````

````python
result1 = pd.merge(
    employees,
    salary, 
    left_on='employee_id', 
    right_on='id')
result1
````

##### 6. **Merge on Multiple Columns**

````python
# Employees
employees = pd.DataFrame({
    "EmpID": [101,101,102,103],
    "Department": ["HR","IT","IT","Sales"],
    "Name": ["Alice","Alice","Bob","Charlie"]
})
# Salary
salary = pd.DataFrame({
    "EmpID": [101,101,102,104],
    "Department": ["HR","Finance","IT","Sales"],
    "Salary": [50000,65000,60000,70000]
})
````

````python
print(employees)
print(salary)
````

````python
result2 = pd.merge(
    employees,
    salary,
    on=['EmpID', 'Department']
)
result2
````

**Notice:**

- (101, HR) matches
- (101, IT) does not match (101, Finance)

Both columns must match.

##### 7. **Add Suffixes for Duplicate Column Names**

Why use suffixes?

Without suffixes, pandas automatically creates *_x* and *_y*, which are less descriptive.

````python
# Employee Details
employees = pd.DataFrame({
    "EmpID": [101,102,103],
    "Department": ["HR","IT","Sales"]
})
# Employee Transfers
transfers = pd.DataFrame({
    "EmpID": [101,102,103],
    "Department": ["Finance","IT","Marketing"]
})
print(employees)
print(transfers)
````

````python
result3 = pd.merge(
    employees,
    transfers,
    on='EmpID',
    suffixes=['_old', '_new']
    )
result3
````

##### 8. **Merge on Index**

````python
# Employees
employees = pd.DataFrame({
    "Name": ["Alice","Bob","Charlie"]
}, index=[101,102,103])
# Salary
salary = pd.DataFrame({
    "Salary": [50000,60000,70000]
}, index=[101,102,104])
````

````python
result4 = pd.merge(
    employees,
    salary,
    left_index=True,
    right_index=True,
    how='outer'
)
result4
````

##### 9. **Useful Merge Indicators**

See where each row came from:

````python
result5 = pd.merge(
    customers,
    orders, 
    on='customer_id', 
    how='outer',
    indicator=True)
result5
````

##### 10. **Joining Three DataFrames**

````python
print(customers)
print(orders)
print(products)
````

````python
result6 = (
    customers
    .merge(orders, on='customer_id', how='left')
    .merge(products, left_on='product', right_on='product_name', how='left')
)
result6
````

##### 11. **Self Join**

A self join means joining a DataFrame with itself.

````python
# Example: Employees and Managers

employees = pd.DataFrame({
    "emp_id": [1, 2, 3, 4],
    "emp_name": ["Alice", "Bob", "Charlie", "David"],
    "manager_id": [None, 1, 1, 2]
})

print(employees)
````

````python
result7 = (
    employees
    .merge(employees, left_on='manager_id', right_on='emp_id', suffixes=('_employee', 'manager'), how='left')
)
result7
````

##### 12. **Cross Join**

A cross join (Cartesian product) combines every row from the first DataFrame with every row from the second DataFrame.


````python
# Example
colors = pd.DataFrame({
    "Color": ["Red", "Blue"]
})

sizes = pd.DataFrame({
    "Size": ["S", "M", "L"]
})
````

````python
result8 = colors.merge(sizes, how='cross')
result8
````

##### **Option 1: *pd.merge()***
```python
result = pd.merge(df1, df2, on='id', how='inner')
```
- This is a function in the pandas module.
- It explicitly shows that you're using pandas to merge two DataFrames.
- Some people prefer it because it looks similar to other pandas functions like pd.concat().

##### **Option 2: *DataFrame.merge()***
```python
result = df1.merge(df2, on='id', how='inner')
```
- This is a method of the DataFrame object.
- It starts from the left DataFrame (df1) and merges df2 into it.
- It is convenient when you want to chain multiple operations.


#### 2. ***join()***

join() is a method used to combine two DataFrames based on their indexes by default.

**Syntax:**
```python
df.join(
    other,
    on=None,
    how='left',
    lsuffix='',
    rsuffix=''
)
```

**Parameters:**
- *other*: DataFrame to join.
- *on='column_name'*: Column in the calling DataFrame to join on the index of the other DataFrame.
- *how='left'*: Type of join.
    - *'left'*
    - *'right'*
    - *'inner'*
    - *'outer'*
    - *'cross'*
- *lsuffix=''*: Suffix for overlapping column names in the left DataFrame.
- *rsuffix=''*: Suffix for overlapping column names in the right DataFrame.

````python

students = pd.DataFrame({
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Kumar']
}, index=[101, 102, 103, 104, 107,])


marks = pd.DataFrame({
    'Marks': [85, 90, 78, 70, 84]
}, index=[101, 102, 104, 105, 106])

print(students)
print(marks)
````

##### 1. **Basic *join()* (Default = Left Join)**

````python
students.join(marks)
````

##### 2. **Outer *join()***

````python
students.join(marks, how='outer')
````

##### 3. **Joining on a Column Using on**

Suppose student IDs are stored as a column instead of the index.


````python
students = pd.DataFrame({
    'StudentID': [101,102,103,104],
    'Name': ['Amit','Priya','Rahul','Sneha']
})

marks = pd.DataFrame({
    'Marks': [85,90,78]
}, index=[101,102,104])
print(students)
print(marks)
````

````python
students.join(marks, on='StudentID')
````

##### 4. **Overlapping Column Names**

Suppose both DataFrames contain a Name column.

````python

students = pd.DataFrame({
    'Name': ['Amit','Priya']
}, index=[101,102])

teachers = pd.DataFrame({
    'Name': ['Raj','Meena']
}, index=[101,102])
print(students)
print(teachers)
````

````python
# Without suffixs:
#students.join(teachers) # ValueError: columns overlap but no suffix specified
# Correct way:
students.join(teachers, lsuffix='_Student', rsuffix='_Teacher')
````

#### 3. ***concat()***

*concat()* is used to combine DataFrames or Series along rows or columns.

**Syntax:**
```python
pd.concat(objs, axis=0, ignore_index=False)
```
**Parameters:**
- objs: List of DataFrames/Series to combine.
- axis=0: Concatenate vertically (rows).
- axis=1: Concatenate horizontally (columns).
- ignore_index=True: Reset the index in the resulting DataFrame.

````python
employees1 = pd.read_csv(r'sample_datasets\merge_join_practice_data\employees_q1.csv')
employees2 = pd.read_csv(r'sample_datasets\merge_join_practice_data\employees_q2.csv')
print(employees1)
print(employees2)
````

##### 1. **Concatenate Rows (Vertical)**

````python
all_emp_v = pd.concat([employees1, employees2])
all_emp_v
````

Reset index:

````python
all_emp_v2 = pd.concat([employees1, employees2], ignore_index=True)
all_emp_v2
````

##### 2. **Concatenate Columns (Horizontal)**

````python
all_emp_h = pd.concat([employees1, employees2], axis=1)
all_emp_h
````

##### 3. **Concatenate DataFrames with Different Columns**

````python
pd.concat([employees,customers])
````

##### 4. **Add Keys to Identify Source DataFrame**

````python
result9 = pd.concat([employees1, employees2], keys=['Team_A', 'Team_B'])
result9
````
