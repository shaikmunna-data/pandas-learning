### **Mathematical Functions in Pandas**

Mathematical functions in Pandas are used to perform calculations on numeric data in a Series or DataFrame. 

They help you summarize, analyze, and manipulate data.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "John"],
    "Age": [25, 30, 35, 28, 32, 27],
    "Salary": [50000, 60000, 70000, 55000, 65000, 50000],
    "Experience": [2, 5, 8, 3, 6, 1]
})

print(df)
```

#### **Aggregation Functions**

- *sum()*       --> Returns the total of values.
- *mean()*      --> Returns the average.
- *median()*    --> Returns the middle value.
- *mode()*      --> Returns the most frequent value(s).
- *min()*       --> Returns the smallest value.
- *max()*       --> Returns the largest value.
- *count()*     --> Counts non-null values.
- *std()*       --> Returns standard deviation.
- *var()*       --> Returns variance.
- *prod()*      --> Multiplies all values together.

- ##### ***sum()***

```python
print(df['Salary'].sum())
```

- ##### ***mean()***

```python
print(df['Salary'].mean())
```

- ##### ***median()***

```python
print(df['Salary'].median())
```

- ##### ***mode()***

```python
print(df['Salary'].mode())
```

- ##### ***min()***

```python
print(df['Salary'].min())
```

- ##### ***max()***

```python
print(df['Salary'].max())
```

- ##### ***count()***

```python
print(df['Salary'].count())
```

- ##### ***std()***

```python
print(df['Salary'].std())
```

- ##### ***var()***

```python
print(df['Salary'].var())
```

- ##### ***prod()***

```python
print(df['Experience'].prod())
```

#### **Cumulative Functions**

- *cumsum()*      --> Running total.
- *cumprod()*     --> Running product.
- *cummax()*      -->Running maximum.
- *cummin()*      -->Running minimum.

- ##### ***cumsum()***

```python
df['Running_Salary'] = df['Salary'].cumsum()
df
```

- ##### ***cumprod()***

```python
print(df['Experience'])
df['Experience'].cumprod()
```

- ##### ***cummax()***

```python
print(df['Salary'])
df['Salary'].cummax()
```

- ##### ***cummin()***

```python
print(df['Salary'])
df['Salary'].cummin()
```

```python
print(df['Salary'])
df['Salary'].pct_change()
```

#### **Other Useful Functions**

- *skew()*          --> Measures data symmetry.
- *kurt()*          --> Measures tail heaviness/outliers.
- *rank()*          --> Assign ranks.
- *diff()*          --> Difference from previous row.
- *pct_change()*    --> Percentage change from previous row.
