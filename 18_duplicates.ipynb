### **Remove Duplicate Rows**

*duplicated()* and *drop_duplicates()* are two closely related pandas methods used to find and remove duplicate rows.

#### 1. **duplicated()**

*duplicated()* checks whether a row is a duplicate of a previous row.

It returns a Boolean Series:

- False → First occurrence
- True → Duplicate occurrence

**Syntax:**
```python
df.duplicated(
    subset=None,
    keep='first'
)
```

**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
</tr>
<tr>
<td><i>subset</i></td>
<td>Column(s) to check duplicates</td>
</tr>
<tr>
<td><i>keep</i></td>
<td>Which duplicate to keep ('first', 'last', False)</td>
</tr>
</table>

````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'David', 'Bob'],
    'Age': [20, 21, 20, 23, 21],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai']
})

print(df)
````

##### **Find duplicate rows**

````python
df.duplicated()
````

##### **Show only duplicate rows**

````python
df[df.duplicated()]
````

##### **Check duplicates using one column**

````python
df.duplicated(subset='Name')
# Only the Name column is checked.
````

##### **Keep last occurrence**

By default *keep='first'*

````python
df.duplicated(keep='last')
````

##### **Mark all duplicates**

Every duplicate row is marked as True.

````python
df.duplicated(keep=False)
````

#### 2. **drop_duplicates()**

Removes duplicate rows.

Instead of returning *True/False*, it returns a new DataFrame.

**Syntax:**
```python
df.drop_duplicates(
    subset=None,
    keep='first',
    inplace=False,
    ignore_index=False
)
```

**Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Description</th>
</tr>
<tr>
<td><i>subset</i></td>
<td>Columns used to identify duplicates</td>
</tr>
<tr>
<td><i>keep</i></td>
<td>'first', 'last', or False</td>
</tr>
<tr>
<td><i>inplace</i></td>
<td>Modify original DataFrame</td>
</tr>
<tr>
<td><i>ignore_index</i></td>
<td>Reset the index in the result</td>
</tr>
</table>

##### **Remove duplicates**

````python
df.drop_duplicates()
````

##### **Remove duplicates using one column**

````python
df.drop_duplicates(subset='Name')
````

##### **Keep last occurrence**

````python
df.drop_duplicates(keep='last')
````

##### **Remove all duplicates**

````python
df.drop_duplicates(keep=False)
````

##### **Reset index**

With *ignore_index=False*

By default *ignore_index* is *False*

````python
df.drop_duplicates()
````

With *ignore_index=True*

````python
df.drop_duplicates(ignore_index=True)
````

##### **Modify the original DataFrame**

````python
print(df)
df.drop_duplicates(ignore_index=True, inplace=True)
df
````
