#### **Problem with Windows paths**

**Suppose you write:**
```python
path = 'C:\newfolder\test.csv'
print(path)
```

**Python sees:**

- \n → New line
- \t → Tab

**Output:**
```
C:
ewfolder    est.csv
```
So the path becomes corrupted.

**Solution: Use raw string (r)**
```python
path = r'C:\newfolder\test.csv'
```

Now Python treats every backslash literally.
This is the most common method.

#### 1. **CSV**

##### **Read CSV**



````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
df
````

````python
df = pd.read_csv(
    r'sample_datasets\Employees.csv',
    header=0,
    names=['emp_id', 'emp_name', 'emp_age', 'emp_dep', 'emp_sal', 'emp_exp'],
    usecols=['emp_id','emp_name', 'emp_sal'],
    index_col='emp_id',
    nrows=10,
    skiprows=1,
    na_values=['Karan Nair'])
df
````
