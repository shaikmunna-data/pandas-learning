### **Pivot Tables**
A pivot table summarizes and aggregates data, similar to Excel's PivotTable.

**Syntax:**
```python
pd.pivot_table(
    data=df,              # DataFrame to summarize
    values='Sales',       # Column to aggregate
    index='Region',       # Row groups
    columns='Category',   # Column groups
    aggfunc='sum',        # Aggregate using sum
    fill_value=0,         # Replace NaN with 0
    margins=True          # Add grand totals
)
```

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Sales.csv')
df
````

#### 1. **Simple Pivot Table**

Average sales by region:


````python
pivot_table1 = pd.pivot_table(
    df,
    index='Region',
    values='Sales'
)
pivot_table1
````

Since no aggfunc is specified, pandas calculates the mean (average) Sales for each Region.

#### 2. **Rows and Columns**

Sales by region and product:

````python
pivot_table2 = pd.pivot_table(
    df,
    index='Region',
    columns='Category',
    values='Sales'
)
pivot_table2
````

#### 3. **Different Aggregations**

Sum sales instead of mean:

````python
pivot_table3 = pd.pivot_table(
    df,
    index='Region',
    columns='Category',
    values='Sales',
    aggfunc='sum'
)
pivot_table3
````

Multiple aggregations:

````python
pivot_table4 = pd.pivot_table(
    df,
    index='Region',
    columns='Category',
    values='Sales',
    aggfunc=['sum', 'mean']
)
pivot_table4
````

#### 4. **Add Totals**

````python
pivot_table5 = pd.pivot_table(
    df,
    index='Region',
    columns='Category',
    values='Sales',
    aggfunc='sum',
    margins=True
)
pivot_table5
````

#### 5. **Multiple Index Levels**

````python
pivot_table6 = pd.pivot_table(
    df,
    index=['Region', 'Category'],
    values='Sales',
    aggfunc='sum',
    margins=True
)
pivot_table6
````

#### 6. **Fill Missing Values**

**without fill_value**

````python
pivot_table8 = pd.pivot_table(
    df,
    index=['Region', 'Category'],
    columns='Product',
    values='Sales',
    aggfunc='sum'
    
)
pivot_table8
````

**with fill_value**

````python
pivot_table9 = pd.pivot_table(
    df,
    index=['Region', 'Category'],
    columns='Product',
    values='Sales',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='Total'
)
pivot_table9
````
