#### 1. **to_csv()**

Exports a DataFrame to a CSV file.

**Syntax:**
```python
df.to_csv(
    path_or_buf,
    index=True,
    columns=None,
    sep=',',
    encoding='utf-8'
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Meaning</th>
<th>Default</th>
</tr>
<tr>
<td><i>path_or_buf</i></td>
<td>File name or file path where the CSV will be saved</td>
<td>Required</td>
</tr>
<tr>
<td><i>index</i></td>
<td>Whether to save the DataFrame index</td>
<td>True</td>
</tr>
<tr>
<td><i>columns</i></td>
<td>Export only specific columns</td>
<td>None (all columns)</td>
</tr>
<tr>
<td><i>sep</i></td>
<td>Separator between values</td>
<td>','</td>
</tr>
<tr>
<td><i>encoding</i></td>
<td>Character encoding</td>
<td>'utf-8'</td>
</tr>
</table>

````python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Class": ["10A", "10A", "10B", "10B", "10A"],
    "Math": [88, 76, 91, 67, 95],
    "Science": [92, 81, 85, 73, 98],
    "Attendance": [95, 88, 97, 90, 99]
})
students
````

````python
students.to_csv(r'sample_datasets\exported_data\students.csv', index=False)
````

#### 2. **to_excel()**

Exports a DataFrame to an Excel file.

**Syntax:**
```python
df.to_excel(
    excel_writer,
    sheet_name='Sheet1',
    index=True
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Meaning</th>
<th>Default</th>
</tr>
<tr>
<td><i>excel_writer</i></td>
<td>Excel file name or path</td>
<td>Required</td>
</tr>
<tr>
<td><i>sheet_name</i></td>
<td>Name of the worksheet</td>
<td>'Sheet1'</td>
</tr>
<tr>
<td><i>index</i></td>
<td>Whether to save the DataFrame index</td>
<td>True</td>
</tr>
</table>

````python
employees = pd.DataFrame({
    "Emp_ID": [1001, 1002, 1003, 1004],
    "Name": ["John", "Sarah", "Mike", "Emma"],
    "Department": ["HR", "IT", "Finance", "Sales"],
    "Salary": [45000, 65000, 55000, 50000],
    "Experience": [3, 6, 4, 2]
})

employees
````

````python
employees.to_excel(r'sample_datasets\exported_data\employees.xlsx', index=False)
````

##### **Multiple Excel Sheets**

Datasets: Employees + Departments + Salaries

````python
employees = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Name": ["John", "Alice", "David"]
})

departments = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Department": ["HR", "IT", "Finance"]
})

salary = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Salary": [45000, 60000, 55000]
})
print(employees)
print(departments)
print(salary)
````

````python
with pd.ExcelWriter(
    r'sample_datasets\exported_data\company.xlsx') as writer:
    employees.to_excel(writer, sheet_name='Employees', index=False)
    departments.to_excel(writer, sheet_name='Departments', index=False)
    salary.to_excel(writer, sheet_name='Salary', index=False)
````

#### 3. **to_json()**

Exports a DataFrame to a JSON file.

**Syntax:**
```python
df.to_json(
    path_or_buf,
    orient='columns',
    indent=None
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Meaning</th>
<th>Default</th>
</tr>
<tr>
<td><i>path_or_buf</i></td>
<td>JSON file name or path</td>
<td>Required</td>
</tr>
<tr>
<td><i>orient</i></td>
<td>Structure of the JSON output</td>
<td>'columns'</td>
</tr>
<tr>
<td><i>indent</i></td>
<td>Pretty-print JSON with indentation</td>
<td>None</td>
</tr>
</table>

**Common *orient* values**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Value</th>
<th>Description</th>
</tr>
<tr>
<td><i>'records'</i></td>
<td>List of dictionaries (most common)</td>
</tr>
<tr>
<td><i>'columns'</i></td>
<td>Dictionary organized by columns (default)</td>
</tr>
<tr>
<td><i>'index'</i></td>
<td>Dictionary organized by index</td>
</tr>
</table>

````python
products = pd.DataFrame({
    "Product_ID": [1, 2, 3, 4],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics"],
    "Price": [65000, 800, 1500, 12000],
    "Stock": [15, 120, 75, 30]
})
products
````

````python
products.to_json(
    r'sample_datasets\exported_data\products.json',
    orient='records',
    indent=4
)
````

#### 4. **to_sql()**

Saves a DataFrame to a SQL database table.

**Syntax:**
```python
df.to_sql(
    name,
    con,
    if_exists='fail',
    index=True
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Meaning</th>
<th>Default</th>
</tr>
<tr>
<td><i>name</i></td>
<td>Name of the SQL table</td>
<td>Required</td>
</tr>
<tr>
<td><i>con</i></td>
<td>Database connection or engine</td>
<td>Required</td>
</tr>
<tr>
<td><i>if_exists</i></td>
<td>What to do if the table already exists</td>
<td>'fail'</td>
</tr>
<tr>
<td><i>index</i></td>
<td>Save the DataFrame index as a column</td>
<td>True</td>
</tr>
</table>

**if_exists values**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Value</th>
<th>Meaning</th>
</tr>
<tr>
<td><i>'fail'</i></td>
<td>Raise an error if the table exists</td>
</tr>
<tr>
<td><i>'replace'</i></td>
<td>Drop the existing table and create a new one</td>
</tr>
<tr>
<td><i>'append'</i></td>
<td>Add rows to the existing table</td>
</tr>
</table>

````python
orders = pd.DataFrame({
    "Order_ID": [5001, 5002, 5003, 5004],
    "Customer": ["Alice", "Bob", "Charlie", "David"],
    "Product": ["Laptop", "Mouse", "Monitor", "Keyboard"],
    "Quantity": [1, 2, 1, 3],
    "Amount": [65000, 1600, 12000, 4500]
})
orders
````

#### 5. **to_parquet()**

Exports a DataFrame to a Parquet file (fast and compressed).

**Syntax:**
```python
df.to_parquet(
    path,
    index=None
)
```

**Important Parameters:**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Parameter</th>
<th>Meaning</th>
<th>Default</th>
</tr>
<tr>
<td><i>path</i></td>
<td>Parquet file name or path</td>
<td>Required</td>
</tr>
<tr>
<td><i>index</i></td>
<td>Whether to save the index</td>
<td>None (handled automatically)</td>
</tr>
</table>

````python
sales = pd.DataFrame({
    "Date": [
        "2025-01-01",
        "2025-01-02",
        "2025-01-03",
        "2025-01-04"
    ],
    "Region": ["North", "South", "East", "West"],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Sales": [120000, 3500, 6000, 25000],
    "Profit": [30000, 900, 1800, 7000]
})
sales
````

````python
sales.to_parquet(r'sample_datasets\exported_data\sales.parquet', index=False)
````

#### 6. **to_pickle()**

Saves a DataFrame as a Python pickle file.


**Syntax:**
```python
df.to_pickle(path)
```

````python
players = pd.DataFrame({
    "Player": ["Virat", "Rohit", "Gill", "KL Rahul"],
    "Team": ["RCB", "MI", "GT", "DC"],
    "Runs": [741, 612, 589, 503],
    "Average": [61.8, 45.3, 49.1, 41.9]
})
players
````

````python
players.to_pickle(r'sample_datasets\exported_data\players.pkl')
````
