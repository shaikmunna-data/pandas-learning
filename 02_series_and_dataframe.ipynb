### **Data Structures**

In **Pandas**, the two most important data structures are
1. Series
2. DataFrame


#### 1. **Series**
A Series is a one-dimensional labeled array that can hold data of any type (integers, strings, floats, etc.)

##### **Creating Series**

```python
import pandas as pd

series1 = pd.Series([10,20, 30, 40, 50])
print(series1)
```

##### **Series with Custom Index**

```python
series2 = pd.Series([10,20,30], index=['a', 'b', 'c'])
print(series2)
```

##### **Accessing Elements**

```python
print(series1[0])
print(series2['a'])
```

#### 2. **DataFrame**

A DataFrame is a two-dimensional labeled data structure with rows and columns, similar to a spreadsheet or SQL table.

##### **Creating a DataFrame**

```python

data = {
    'Name': ['Santosh', 'Ali', 'Rahul'],
    'Age': [25, 30, 35],
    'City': ['Mumbai', 'Hydrabad', 'Delhi']
}

df = pd.DataFrame(data)
#
print(df)
```

##### **Accessing Columns**

```python
print(df['Name'])
```

```python
print(df['City'])
```

##### **Accessing Rows**

```python
print(df.loc[0])   # By label
print(df.iloc[0])  # By position
```

##### **Relationship Between Series and DataFrame**

A DataFrame can be thought of as a collection of Series sharing the same index.

```python
print(type(df['Name']))
```
