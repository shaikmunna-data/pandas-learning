### **Data Type Conversion**

```python
import pandas as pd

df = pd.DataFrame({
    'age': ['25', '30', '35'],
    'salary': ['50000', '60000', '70000'],
    'joined': ['2024-01-10', '2024-03-15', '2024-06-20'],
    'active': ['True', 'False', 'True']
})
```

#### 1. **Check current data types**

```python
df.dtypes
```

#### 2. **Convert using astype()**

**String → Integer**

```python
df['age'] = df['age'].astype(int)
df['age'].dtypes
```

**String → Float**

```python
df['salary'] = df['salary'].astype(float)
df['salary'].dtype
```

**Multiple columns together**

```python
df = df.astype({
    'age':'int64',
    'salary':'float64'
})
```

```python
df.dtypes
```

#### 3. **Convert to datetime**

Use pd.to_datetime().

```python
df['joined'] = pd.to_datetime(df['joined'])
print(df['joined'].dtypes)
```

#### 4. **Convert to boolean**

```python
df['active'] = df['active'].map({
    'True': True,
    'False': False
})

df['active'].dtype
```

#### 5. **Convert to category (memory efficient)**



```python
dept = pd.DataFrame({
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'HR']
})

print(dept)
print(dept.dtypes)
```

**Convert to category**

```python
dept['Department'] = dept['Department'].astype('category')
print(dept.dtypes)
dept
```

#### 6. **pd.to_numeric() — safe numeric conversion**

Use it when you have a column that should contain numbers, but may contain invalid values like "abc", "N/A", "-", etc.

```python

# Sample dataset
emp = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': ['25', '30', '28', '35'],          # strings
    'Salary': ['50000', '65000', '55000', '72000'],  # strings
    'Bonus': ['1000', 'abc', '1500', None]    # mixed values
})

print(emp)
emp.dtypes
```

**Convert Age and Salary**

```python
emp['Age'] = pd.to_numeric(emp['Age'])
emp['Salary'] = pd.to_numeric(emp['Salary'])
emp.dtypes
```

What happened?

- '25' → 25

- '50000' → 50000

- Pandas chose int64 because all values are whole numbers.

**Convert a column with invalid values**

The Bonus column has 'abc', which is not a number.

Without error handling

```python
# pd.to_numeric(emp['Bonus'])
# ValueError: Unable to parse string 'abc'
```

**Use errors='coerce'**

```python
emp['Bonus'] = pd.to_numeric(emp['Bonus'], errors='coerce')
print(emp)
emp.dtypes
```

Why float64?

Because:

- valid numbers become 1000.0, 1500.0

- invalid values become NaN

- NaN is a floating-point value, so pandas converts the whole column to float64.
