### **GroupBy in Pandas**

*groupby()* is one of the most important operations in pandas. It is used to group rows that have the same value and then perform calculations on each group.

Think of it as:

Split → Apply → Combine

1. Split data into groups
2. Apply a function (sum, mean, count, etc.)
3. Combine the results

**General Syntax**
```python
df.groupby(group_column)[value_column].operation()
```

````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv')
df
````

#### 1. **Basic *groupby()***

**Example:**

```python
df.groupby("Department")["Salary"].sum()
```

**Meaning:**

1. Group employees by Department.
2. Take Salary column.
3. Add salaries in each department.

````python
df.groupby('Department')['Salary'].sum()
````

#### 2. **View Groups**

Shows row indexes in each group.

````python
df.groupby('Department').groups
````

- *mean()*
- *min()*
- *max()*
- *count()*
- *median()*
- *first()*
- *last()*

- ##### ***sum()***

````python
df.groupby('Department')['Salary'].sum()
````

````python
df.groupby('Department')['Salary'].min()
````

#### 4. **Multiple Aggregations**

**Instead of running many queries:**

``` python
df.groupby("Department")["Salary"].sum()

df.groupby("Department")["Salary"].mean()

df.groupby("Department")["Salary"].max()
```

**Use *agg()*:**

```python
df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "min", "max", "count"]
```

````python
df.groupby('Department')['Salary'].agg(['sum', 'max', 'min', 'count', 'mean', 'median'])
````

#### 5. ***as_index=False***

**Default:**

```python
df.groupby("Department")["Salary"].sum()
```

Department becomes index.

**To keep it as a normal column:**
```python
df.groupby(
    "Department",
    as_index=False
)["Salary"].sum()
```

````python
df.groupby('Department', as_index=False)['Salary'].sum()
````

#### 6. **Get Particular Group**

````python
df.groupby('Department').get_group('IT')
````

#### 7. **Sorting Groups**

**Default:**

```python
df.groupby("Department").sum()
```
Groups are sorted alphabetically.

**Disable sorting:**

```python
df.groupby(
    "Department",
    sort=False
).sum()
```

````python
df.groupby('Department', sort=False)['Salary'].sum()
````

#### 8. **Group By Multiple Columns**

````python
new_df = pd.DataFrame({
    "EmpID": [101,102,103,104,105,106,107,108,109,110],
    "Name": ["Ravi","Priya","Kiran","Anu","Raj",
             "Deepa","Arjun","Sneha","Vijay","Pooja"],
    "Department": ["IT","HR","IT","HR","Sales",
                   "Sales","IT","HR","Sales","IT"],
    "City": ["Hyderabad","Chennai","Bangalore","Chennai",
             "Mumbai","Mumbai","Hyderabad","Bangalore",
             "Mumbai","Bangalore"],
    "Gender": ["M","F","M","F","M",
               "F","M","F","M","F"],
    "Salary": [50000,40000,60000,45000,55000,
               65000,70000,48000,52000,62000],
    "Bonus": [5000,3000,7000,3500,6000,
              8000,9000,4000,5500,7500]
})
new_df
````

````python
new_df.groupby(['Department', 'City'])['Salary'].sum()
````

#### 9. **Different Aggregations for Different Columns**

**Suppose:**

- Salary → average salary
- Bonus → total bonus
- EmpID → employee count

**Use a dictionary:**

```python
df.groupby("Department").agg({
    "Salary": "mean",
    "Bonus": "sum",
    "EmpID": "count"
})
```

````python
new_df.groupby('Department').agg({
    'Salary':'mean',
    'Bonus':'sum',
    'EmpID':'count'
})
````

#### 10. **Multiple Aggregations on Multiple Columns**

````python
new_df.groupby('Department').agg({
    'Salary':['sum', 'mean', 'max', 'min'],
    'Bonus':['sum', 'mean', 'max', 'min']
})
````

#### 11. **Named Aggregations**

````python
new_df.groupby('Department').agg(
    avg_salary = ('Salary', 'mean'),
    total_bonus = ('Bonus','sum'),
    no_of_employees = ('EmpID', 'count')
)
````

#### 12. **Group by Multiple Value Columns**

````python
new_df.groupby('Department')[['Salary', 'Bonus']].sum() # Salary & Bonus summed for each department
````

#### 13. **Transform**

Returns the same number of rows as original DataFrame.

````python
new_df['Max Salary'] = new_df.groupby('Department')['Salary'].transform('max')
````

````python
new_df.sort_values(by=['Department', 'Salary'], ascending=[1,0])
````

#### **Why do we need Transform?**

**Suppose you want to know:**

What is the average salary of each employee’s department?

**If you use agg():**


````python
new_df.groupby('Department')['Salary'].agg('mean')
````


You get only 3 rows, not 10.

But we want the average salary attached to every employee row.

**Using transform()**

````python
new_df['Dept_avg_sal'] = new_df.groupby('Department')['Salary'].transform('mean')
````

````python
new_df.sort_values(by='Department')
````
