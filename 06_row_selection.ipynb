### **Row Selection in Pandas**

Row selection means choosing specific rows from a DataFrame.

#####  ***loc* and *iloc* in Pandas**

*loc* and *iloc* are the two main indexing methods in Pandas. They are used to select rows, columns, or both.

*************************************************************************************************************
1. ***iloc* (Integer Position-Based Indexing)**

*iloc* selects data using integer positions.

**Syntax:**

```python
df.iloc[row_position, column_position]
```
*************************************************************************************************************
2. ***loc* (Label-Based Indexing)**

*loc* selects data using row labels (index names) and column labels (column names).

**Syntax:**

```python
df.loc[row_label, column_label]
```
*************************************************************************************************************


````python
import pandas as pd
df = pd.read_csv(r'sample_datasets\Employees.csv', index_col='EmpID')
df
````

#### 1. **Select Rows by Position**

- Using *iloc*

1. **Select a Single Row using *iloc***

````python
df.iloc[1]
````

2. **Select Multiple Rows using *iloc***

````python
df.iloc[[0,2]]
````

3. **Select a Range of Rows using iloc**

````python
df.iloc[0:3]
````

#### 2. **Select Rows by Label**
- Using *loc*

1. **Select a Single Row using *loc***

````python
df.loc[1005]
````

2. **Select Multiple Rows using *loc***

````python
df.loc[[1005,1010]]
````

3. **Select a Range of Rows using *loc***

````python
df.loc[1015:1020]
````

#### 3. **Select First and Last Row**

````python
df.iloc[[0,-1]]
````

#### 4. **Select Specific Rows and Columns**

````python
df.iloc[5:10, 0:4]
````

````python
df.iloc[[1,4],[1,4]]
````

````python
df.loc[[1001,1010], ['Department','Name','Experience']]
````

````python
df.loc[1004:1010, 'Name':'Department']
````
