### **Pandas Introduction in Python**
Pandas is one of the most popular Python libraries used for data analysis and data manipulation. It helps you work with tables, spreadsheets, CSV files, Excel files, and large datasets easily.

Think of Pandas as Excel inside Python, but much more powerful.

#### **Why do we use Pandas?**

Without Pandas, handling large amounts of data in Python is difficult.

Pandas makes it easy to:

- ✅ Read CSV and Excel files
- ✅ Store data in table format
- ✅ Select rows and columns
- ✅ Filter data
- ✅ Sort data
- ✅ Handle missing values
- ✅ Perform calculations
- ✅ Analyze and summarize data

#### **Importing Pandas**
```python
import pandas as pd
```

Here:
- import → Imports the library.
- pandas → Library name.
- as pd → Gives it a short name.
